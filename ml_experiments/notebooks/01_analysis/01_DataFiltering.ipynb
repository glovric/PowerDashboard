```python
import pandas as pd
```

# Load raw time series data

```python
path = "../../data/time_series_60min_singleindex.csv"
df = pd.read_csv(path)

df.head()
```

# Extract countries

```python
countries = set(c[:2].upper() for c in df.columns if "timestamp" not in c) # Create set of countries (no duplicates)
countries.remove("GB")
countries.add("GB_GBN") # GB column is actually GB_GBN
countries = list(countries)
countries.sort()
print("|".join(countries))
```

# Filter only timestamp column and load columns

```python
filtered_columns = ["utc_timestamp"] + [f'{c}_load_actual_entsoe_transparency' for c in countries]
df = df[filtered_columns]
df.rename(columns={"GB_GBN_load_actual_entsoe_transparency": "GB_load_actual_entsoe_transparency"}, inplace=True)
df.head()
```
